## 1. What is Social Listening?

Social listening and social monitoring are related but distinct practices:

| Dimension | Social Monitoring | Social Listening |
|-----------|------------------|------------------|
| **Focus** | Direct mentions, tags, replies | Broad conversations, even without tagging |
| **Orientation** | Reactive — respond to events | Proactive — extract strategic insight |
| **Data type** | Primarily quantitative | Qualitative + Quantitative |
| **Output** | Incident mgmt, customer service | Brand strategy, product direction, competitive positioning |
| **Time horizon** | Real-time / operational | Longitudinal / strategic |

### Classic Example: McDonald's "$18 Big Mac" (2024)
- **Monitoring** caught individual complaints → triggered CS responses
- **Listening** revealed a systemic affordability concern → led to a $5 meal deal launch and public commitment to slow price increases

## 2. Key Metrics

### 2.1 Reach & Impressions
- **Reach**: unique accounts that *could have* seen content (audience size)
- **Impressions**: total times content was displayed, including repeat views
- Ratio: one viral piece can have 10× impressions vs. reach due to algorithmic re-surfacing

### 2.2 Engagement Metrics
```
Engagement Rate = (likes + comments + shares + saves) / impressions
```
Platform-specific variants:
- **Reddit**: upvote ratio, comment depth
- **YouTube**: watch time %, comment sentiment
- **HackerNews**: point velocity (points per hour)

### 2.3 Sentiment (Net Sentiment Score)
```
NSS = ((positive_mentions - negative_mentions) / total_mentions) × 100
Range: -100 (all negative) to +100 (all positive)
```

### 2.4 Share of Voice (SOV)
```
SOV = (your_brand_mentions / total_industry_mentions) × 100
```
Example: 500 brand mentions in a category with 10,000 total = 5% SOV

### 2.5 Brand Health Score (Composite)
| Component | Weight (example) | Measure |
|-----------|-----------------|--------|
| Awareness | 20% | Mention volume |
| Sentiment | 35% | NSS rolling 30-day |
| SOV | 20% | vs. top 3 competitors |
| Advocacy | 15% | Organic positive mentions |
| Review Score | 10% | Weighted avg. platform ratings |

## 3. Virality Analysis

### 3.1 Viral Coefficient (K-Factor)
```
K = (invitations_sent_per_user) × (conversion_rate)

K > 1  → Exponential growth
K = 1  → Steady state  
K < 1  → Decay
```

**Key 2024 finding**: K-factor is a *growth* metric, not a quality metric. High virality does not always enhance meaningful engagement.

### 3.2 Cascade Analysis
A diffusion cascade maps how content spreads from origin through resharing hops.

```
Origin post
    ├── Share A (hop 1)
    │       ├── Share A1 (hop 2)
    │       └── Share A2 (hop 2)
    ├── Share B (hop 1)
    └── Share C (hop 1)
            └── Share C1 (hop 2)
                    └── Share C1a (hop 3)

Cascade Depth = 3 (longest chain)
Cascade Width = 3 (max breadth at hop 1)
```

**State-of-the-art**: Temporal-Spatial Cascade Convolutional Networks (ViralGCN) predict viral spread from early cascade structure.

**Critical insight from research**: *Community spread diversity* predicts virality better than raw share count. Content that penetrates multiple distinct communities early is far more likely to go viral.

### 3.3 Velocity Measurement
```python
# Share velocity = shares per hour since publish
# Z-score spike detection for anomaly alerting
# if z_score > 3.0: trigger viral alert
```

### 3.4 Information Diffusion Models
Borrowed from epidemiology:

| Model | States | Best For |
|-------|--------|----------|
| **SI** | Susceptible → Infected | Content that spreads indefinitely |
| **SIR** | S → I → Recovered | Memes/trends that fade |
| **SEIR** | S → Exposed → I → R | Realistic latency before sharing |
| **SEIHR** (2024) | + Hesitant state | Human attitude heterogeneity |

Python library: `NDlib` — implements all models on NetworkX graphs.

## 4. Network & Connection Analysis

### 4.1 Graph Fundamentals for Social Media
```
Nodes = users, posts, hashtags, communities
Edges = follows, retweets, replies, co-mentions
Direction = who influenced whom
Weight = strength of connection (interaction frequency)
```

### 4.2 Centrality Metrics (Influencer Identification)

| Metric | What It Measures | Best For |
|--------|-----------------|----------|
| **Degree Centrality** | Raw connection count | Broadly connected nodes |
| **Betweenness Centrality** | Bridge between communities | Information brokers |
| **Eigenvector Centrality** | Connected to well-connected | True influence ("quality" of network) |
| **PageRank** | Iterative eigenvector | Same as above, more stable |
| **Closeness** | Average distance to all nodes | Speed of information spread |

```python
import networkx as nx

G = nx.DiGraph()  # directed graph (A retweeted B)
nx.pagerank(G)               # PageRank scores
nx.betweenness_centrality(G) # Bridge identifiers
nx.eigenvector_centrality(G) # True influence
```

### 4.3 Community Detection

| Algorithm | Best For | Scale |
|-----------|----------|-------|
| **Louvain** | General purpose, modularity-maximizing | Millions of nodes |
| **Girvan-Newman** | Hierarchical, bridge-based | Small graphs (<10K) |
| **INFOMAP** | Directed networks (retweet graphs) | Large directed graphs |
| **Label Propagation** | Fast approximate | Very large graphs |

```python
import community as community_louvain  # python-louvain
partition = community_louvain.best_partition(G)
# Returns {node: community_id} dict
```

### 4.4 Practical Applications
1. **Echo chamber detection**: communities with high internal density but low cross-community edges
2. **Influence path tracing**: how did a narrative spread from fringe communities to mainstream?
3. **Coordinated behavior detection**: abnormally synchronized posting patterns in the graph
4. **Audience overlap**: which communities share audiences with your brand?

### 5.5 Aspect-Based Sentiment Analysis (ABSA)

Extracts *which features* have positive/negative sentiment:
```
Input:  "The battery life is terrible but the camera is stunning"
Output: battery_life → NEGATIVE
        camera       → POSITIVE
```

Four subtasks:
1. **ATE** — Aspect Term Extraction: find "battery life", "camera"
2. **OTE** — Opinion Term Extraction: find "terrible", "stunning"
3. **ASC** — Aspect Sentiment Classification: assign polarity
4. **ACD** — Aspect Category Detection: {Hardware, Power} vs {Camera}

```python
from pyabsa import ATEPCCheckpointManager
aspextractor = ATEPCCheckpointManager.get_aspect_extractor(
    checkpoint="english", auto_device=True
)
result = aspextractor.extract_aspect(inference_source=["The battery is poor but screen is great"])
```

### 7.2 Semi-Structured Nature of Reviews
```
STRUCTURED           SEMI-STRUCTURED      UNSTRUCTURED
────────────         ───────────────      ────────────
Star rating (1-5)    Review title         Review body text
Date                 Verified purchase    Tips/suggestions
Helpful votes        Category tags        Photos
```

**Advantage over social**: explicit product context + numerical anchor (star rating)  
**Disadvantage vs social**: lower velocity, selection bias (motivated customers write reviews)

### 7.3 Review Analysis Techniques

**Aspect Extraction Pipeline:**
```
Review Text
    → Tokenize & Parse (spaCy dependency parsing)
    → Extract noun-adjective pairs
    → Cluster aspects by semantic similarity
    → Assign sentiment per aspect (ABSA)
    → Aggregate across all reviews
    → Output: {feature: sentiment_score} dict
```

**Fake Review Detection Signals:**
1. **Behavioral**: reviewer posting frequency, timestamp clustering
2. **Textual**: generic language, aspect repetition, extreme polarity without specifics
3. **Network**: co-reviewer bipartite graph — suspicious clusters of reviewers sharing the same products
4. **2024 method**: ABSA + Graph Convolutional Networks detects coordinated fake review campaigns

## 8. Integrating Social + Reviews

### 8.1 Complementary Signal Map

| Dimension | Social Media | Online Reviews |
|-----------|-------------|----------------|
| **Velocity** | High (real-time) | Low (deliberate) |
| **Depth** | Low (short posts) | High (detailed) |
| **Context** | Broad (any topic) | Narrow (specific product) |
| **Emotion** | Spontaneous | Considered |
| **Virality signal** | Yes | No |
| **Purchase intent** | Weak | Strong (verified buyers) |
| **Structured anchor** | No | Yes (star rating) |

**Combined insight**: Social captures the leading edge of opinion shift (real-time); reviews capture the trailing, considered verdict. Together = complete picture across the customer journey.

### 8.2 Unified Pipeline Architecture

```
DATA INGESTION
──────────────
Reddit (PRAW)      ─┐
HackerNews API     ─┤
GDELT News         ─┤──▶ Queue/Buffer ──▶ NLP Processing ──▶ Data Store
Mastodon API       ─┤
Amazon Reviews     ─┤
Yelp Reviews       ─┘

NLP PROCESSING
──────────────
1. Language detection (fasttext)
2. Sentiment (VADER → Twitter-RoBERTa)
3. Emotion (NRC Lexicon / GoEmotions)
4. Topic modeling (BERTopic)
5. NER (spaCy + custom brand rules)
6. ABSA (pyabsa)
7. Anomaly detection (Prophet + IsolationForest)

METRICS LAYER
─────────────
• Net Sentiment Score (7-day, 30-day rolling)
• Share of Voice vs. competitors
• Topic Distribution
• Aspect-Level Satisfaction
• Crisis Score (anomaly output)
• Review Composite Score
• Virality Index (K-factor proxy)
```
